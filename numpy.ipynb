```python
import numpy as np

x= np.array([2,3])
z= np.array([4,1])

print("Addition:", x+z)
print("Subtraction: ", x-z)
print("Scaler Mutiplication: ", 3*x)
print("Element wise Multiplication: ", x*z)
```

```python
import numpy as np

x= np.array([5.0, 7.0])

w= np.array([6.0, 2.0])
b= 10.0

prediction= np.dot(w, x)+b

print("Predicted Score: ", prediction)
```

```python
import numpy as np

# Transpose
A = np.array([[1, 2, 3],
              [4, 5, 6]])

print(A.T.shape)  
```

```python
# Inverse of a nonsingular matrix
B = np.array([[2.0, 0.0],
              [0.0, 4.0]])
print(np.linalg.inv(B))
```

```python
# Rank
C = np.array([[1, 2],
              [2, 4]])

print(np.linalg.matrix_rank(C))
```

```python
# Projection of x onto the direction u
x = np.array([3.0, 4.0])
u = np.array([1.0, 0.0])

projection = (np.dot(x, u) / np.dot(u, u)) * u
print(projection)
```
