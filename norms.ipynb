```python
import numpy as np

A= np.array([2, 3])
B= np.array([5, 7])

# Norms of one vector
print(np.linalg.norm(A, ord=1)) # Manhattan
print(np.linalg.norm(A, ord=2)) # Euclidean
print(np.linalg.norm(A, ord= np.inf))  # max

#Distance bw two vectors
print(np.linalg.norm(A-B))   # euclidean
print(np.sum(np.abs(A-B)))   #manhattan
```
